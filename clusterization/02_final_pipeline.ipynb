### Dimesion redusing

```python
X = sparse.load_npz("../datasets/clusterization/train.npz")
M, N = X.shape
# масштабируем
scaler = RobustScaler(with_centering=False)
X = scaler.fit_transform(X)
# PCA
pca_new = PCA(n_components=40)
X_embedded_pca = pca_new.fit_transform(X)
```
