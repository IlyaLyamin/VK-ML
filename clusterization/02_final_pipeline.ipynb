```python
from scipy import sparse
import numpy as np
import pandas as pd
import os

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.mixture import GaussianMixture

from tools import plot_3D

RANDOM_STATE = 42

os.environ["MKL_THREADING_LAYER"] = "GNU"
```

### Dimesion redusing

```python
X = sparse.load_npz("../datasets/clusterization/train.npz")
X = X.toarray()
M, N = X.shape
# масштабируем
scaler = StandardScaler(with_mean=False)
X_scaled = scaler.fit_transform(X)
# PCA
pca_new = PCA(n_components=40)
X_embedded_pca = pca_new.fit_transform(X_scaled)
```

### Clustering

```python
model = GaussianMixture(n_components=10,
                        covariance_type="full",
                        n_init=3,
                        random_state=RANDOM_STATE,
                        init_params="kmeans")
labels = model.fit_predict(X_embedded_pca)
```

```python
sbm = pd.DataFrame({"ID": np.arange(labels.size), "TARGET": labels})
sbm.to_csv("final_submission.csv", index=False)
```

```python
plot_3D(X_embedded_pca, labels, reduce_dim=True, save_html=True)
```
